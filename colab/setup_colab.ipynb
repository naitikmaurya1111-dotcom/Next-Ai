## Start Server & Tunnel
Replace `YOUR_NGROK_TOKEN` with your actual token from https://dashboard.ngrok.com/get-started/your-authtoken

```python
import asyncio
from pyngrok import ngrok
import uvicorn
import subprocess
import qrcode
from IPython.display import display

# Set your ngrok token
NGROK_TOKEN = 'YOUR_NGROK_TOKEN' # <-- Replace this
ngrok.set_auth_token(NGROK_TOKEN)

# Start tunnel
tunnel = ngrok.connect(8000, 'http')
ws_url = tunnel.public_url.replace("https", "wss").replace("http", "ws") + "/ws"

print(f'AGY Chat Server HTTP URL: {tunnel.public_url}')
print(f'\nWebSocket URL: {ws_url}')
print('Copy the WebSocket URL into your Android app Settings!')

# Generate QR Code for easy scanning
qr = qrcode.QRCode(version=1, box_size=10, border=5)
qr.add_data(ws_url)
qr.make(fit=True)
img = qr.make_image(fill_color="black", back_color="white")
display(img)

# Start server
subprocess.Popen(['uvicorn', 'main:app', '--host', '0.0.0.0', '--port', '8000'])
```
